````python
pytrends = TrendReq()

# 1. Google Trends bağlantısını başlatıyoruz
# hl='en-US' -> Arama sonuçlarının dilini İngilizce ayarlar
# tz=360     -> Saat dilimini (Timezone) ayarlar
pytrend = TrendReq(hl='en-US', tz=180)

# 2. İncelemek istediğimiz kelimeleri liste olarak tanımlıyoruz
x= ["Java", "Python", "C++"]

# 3. Parametreleri(istek) Google Trends'e gönderiyoruz (Payload oluşturma)
pytrend.build_payload(kw_list= x, timeframe="today 5-y", geo= "TR")
````

# Df' i kopyalayıp eksiltelim ve linner interpolationı test edelim
gerçek pytrends verisindeki gerçek bir değeri bilerek sileceğiz → interpolation ile tahmin edeceğiz → gerçek değerle tahmini karşılaştıracağız

````python
test_df= df.copy()
````

````python
help(df.copy)
````

````python
import numpy as np

# 1) Orijinal veriyi bozmamak için kopyasını oluştur
test_df = df.copy()

# 2) Test edeceğimiz gerçek değeri sakla
gerçek_deger = test_df.loc["2024-05-05", "Python"]

# 3) O değeri bilerek eksik hale getir
test_df.loc["2024-05-05", "Python"] = np.nan

# 4) Eksik halini göster
print("=== INTERPOLATION ÖNCESİ ===")
print(test_df.loc["2024-04-28":"2024-05-12", "Python"])

# 5) Python sütunundaki eksik değerleri linear interpolation ile doldur
test_df["Python"] = test_df["Python"].interpolate(method="linear")

# 6) Doldurulmuş halini göster
print("\n=== INTERPOLATION SONRASI ===")
print(test_df.loc["2024-04-28":"2024-05-12", "Python"])

# 7) Interpolation'ın tahmin ettiği değeri al
tahmin = test_df.loc["2024-05-05", "Python"]

# 8) Gerçek değer ile tahmini karşılaştır
print("\n=== SONUÇ ===")
print("Gerçek değer:", gerçek_deger)
print("Tahmin edilen:", tahmin)
print("Hata:", abs(gerçek_deger - tahmin))
````
